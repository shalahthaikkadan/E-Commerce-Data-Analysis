```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

# Load the dataset
df = pd.read_csv("Dataset for Data Analytics - Sheet1.csv")

# Show first 5 rows
df.head()
```

```python
# Dataset information
df.info()

# Statistical summary
df.describe()

# Number of rows and columns
print("Dataset Shape:", df.shape)

# Column names
print("\nColumns:")
print(df.columns)

# Missing values
print("\nMissing Values:")
print(df.isnull().sum())
```

```python
plt.figure(figsize=(12,5))

sns.heatmap(
    df.isnull(),
    cbar=False,
    cmap="viridis"
)

plt.title("Missing Values Heatmap")

plt.show()
```

```python
# Fill Shipping Address with "Unknown"
df["ShippingAddress"] = df["ShippingAddress"].fillna("Unknown")

# Fill Tracking Number
df["TrackingNumber"] = df["TrackingNumber"].fillna("Not Available")

# Fill Coupon Code
df["CouponCode"] = df["CouponCode"].fillna("No Coupon")

# Fill Referral Source
df["ReferralSource"] = df["ReferralSource"].fillna("Unknown")

print(df.isnull().sum())
```

```python
def detect_outliers(column):

    Q1 = df[column].quantile(0.25)

    Q3 = df[column].quantile(0.75)

    IQR = Q3 - Q1

    lower = Q1 - 1.5 * IQR

    upper = Q3 + 1.5 * IQR

    outliers = df[
        (df[column] < lower) |
        (df[column] > upper)
    ]

    print(column)
    print("Lower:", lower)
    print("Upper:", upper)
    print("Outliers:", len(outliers))
    print("-----------------------")

detect_outliers("Quantity")

detect_outliers("UnitPrice")

detect_outliers("ItemsInCart")

detect_outliers("TotalPrice")
```

```python
def cap_outliers(column):

    Q1 = df[column].quantile(0.25)

    Q3 = df[column].quantile(0.75)

    IQR = Q3 - Q1

    lower = Q1 - 1.5 * IQR

    upper = Q3 + 1.5 * IQR

    df[column] = df[column].clip(lower, upper)

cap_outliers("Quantity")

cap_outliers("UnitPrice")

cap_outliers("ItemsInCart")

cap_outliers("TotalPrice")

print("Outliers Treated Successfully")
```

```python
# Average Price Per Item
df["AveragePricePerItem"] = (
    df["TotalPrice"] / df["Quantity"]
)

# Coupon Used
df["CouponUsed"] = (
    df["CouponCode"] != "No Coupon"
).astype(int)

# Convert Date
df["Date"] = pd.to_datetime(df["Date"])

# Month
df["OrderMonth"] = df["Date"].dt.month_name()

# Weekday
df["Weekday"] = df["Date"].dt.day_name()

df.head()
```

```python
df_encoded = pd.get_dummies(

    df,

    columns=[

        "Product",

        "PaymentMethod",

        "OrderStatus",

        "ReferralSource",

        "OrderMonth",

        "Weekday"

    ],

    drop_first=True

)

df_encoded.head()
```

```python
plt.figure(figsize=(15,10))

sns.heatmap(

    df_encoded.corr(numeric_only=True),

    cmap="coolwarm"

)

plt.title("Correlation Matrix")

plt.show()
```

```python
df_encoded.to_csv(

    "Cleaned_Ecommerce_Dataset.csv",

    index=False

)

print("Project Completed Successfully")

print("Clean dataset saved as")

print("Cleaned_Ecommerce_Dataset.csv")
```
